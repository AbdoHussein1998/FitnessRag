```python
from bson import ObjectId
from qdrant_client.http.models import PointStruct

def new_func_upsert(record_id, vector, text, metadata):
    # Convert Mongo ObjectId string to int for Qdrant
    record_int = int(ObjectId(record_id).binary.hex(), 16) % (2**63)

    point = PointStruct(
        id=record_int,
        vector=vector,
        payload={
            "text": text,
            "metadata": metadata
        }
    )

    # Upsert point to Qdrant collection
    return qdrant_client.upsert(
        collection_name="chunks",
        points=[point]
    )


# Example usage:
doc = db_databse["chunks"].find_one({})

# Ensure embedding is defined and is a list of floats of length 1536
embedding = model.encode(doc['text']).tolist()
print(len(embedding))

# qdrant_client.delete_collection(collection_name="chunks")
qdrant_client.create_collection(
    collection_name="chunks",
    vectors_config={"vector": {"size": len(embedding), "distance": "Cosine"}}
)

```

```python
from bson import ObjectId
from qdrant_client.http.models import PointStruct

def new_func_upsert(record_id, vector, text, metadata):
    record_int = int.from_bytes(ObjectId(record_id).binary, byteorder="big") % (2**63)

    point = PointStruct(
        id=record_int,
        vector=vector,
        payload={
            "text": text,
            "metadata": metadata
        }
    )

    return qdrant_client.upsert(
        collection_name="chunks",
        points=[point]
    )


# Example usage:
doc = db_databse["chunks"].find_one({})
embedding = model.encode(doc['text']).tolist()

# Ensure clean collection setup
qdrant_client.recreate_collection(
    collection_name="chunks",
    vectors_config={"vector": {"size": len(embedding), "distance": "Cosine"}}
)

new_func_upsert(
    record_id=str(doc["_id"]),
    vector={"vector": embedding},  # ✅ name must match the one in create_collection
    text=doc["text"],
    metadata={"file_id": doc["file_id"], "title": doc["title"]}
)
```
