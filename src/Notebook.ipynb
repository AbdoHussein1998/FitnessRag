```python
async def fetch_in_batches(db_collection, batch_size=1000, sleep_time=3):
    cursor = db_collection.find()
    batch = []
    counter = 0

    async for doc in cursor:
        batch.append(doc)

        if len(batch) == batch_size:
            counter += len(batch)
            yield batch  # process this batch outside
            batch = []
            print(f"Sleeping {sleep_time}s after {counter} docs")
            await asyncio.sleep(sleep_time)

    # yield remaining docs
    if batch:
        counter += len(batch)
        yield batch
        print(f"Final batch processed, total {counter} docs")
```
